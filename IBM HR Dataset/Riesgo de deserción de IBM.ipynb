Se procede a idenetificar y eliminar variables que redundantes o que no parte informacion:

```python
clean_IBM = pd.DataFrame()
columns_name = list(IBM_df.columns)
cleaned_variables = []

for column_name in columns_name:
    column = IBM_df[column_name]

    #Check unique values
    unique_values = column.unique()
    unique_values = unique_values[np.invert(pd.isna(unique_values))]
    unique_coverage = len(unique_values)/len(column)

    #Count nulls
    nan_counts = column.isna().sum()
    coverage = (len(column) - nan_counts)/len(column)

    if not((len(unique_values) <= 1) or (unique_coverage == 1) or (coverage < 0.1)):

        #Object columns cleaning
        column = column.map(lambda value : value.replace('\n', ' ') if type(value)==str else value)

        clean_IBM = pd.concat([clean_IBM, column], axis = 1)
    else:
        cleaned_variables.append(column_name)

clean_IBM.head()
```

```python
# Definir un diccionario que contenga los valores categóricos para cada columna
mappings = {
    'Education': {1: 'Below College', 2: 'College', 3: 'Bachelor', 4: 'Master', 5: 'Doctor'},
    'EnvironmentSatisfaction': {1: 'Low', 2: 'Medium', 3: 'High', 4: 'Very High'},
    'JobInvolvement': {1: 'Low', 2: 'Medium', 3: 'High', 4: 'Very High'},
    'JobSatisfaction': {1: 'Low', 2: 'Medium', 3: 'High', 4: 'Very High'},
    'PerformanceRating': {1: 'Low', 2: 'Good', 3: 'Excellent', 4: 'Outstanding'},
    'RelationshipSatisfaction': {1: 'Low', 2: 'Medium', 3: 'High', 4: 'Very High'},
    'WorkLifeBalance': {1: 'Bad', 2: 'Good', 3: 'Better', 4: 'Best'},
    'JobLevel': {1: "Entry Level", 2: "Junior Level", 3: "Mid Level", 4: "Senior Level", 5: "Executive Level"}
}

# Función para reemplazar valores numéricos por categóricos
def reemplazar_valores_categoricos(dataset):
    for column, mapping in mappings.items():
        if column in dataset.columns:
            dataset[column] = dataset[column].map(mapping)
    return dataset

# Aplicar la función en tu dataset 'exploratory data'
exploratory_analysis = reemplazar_valores_categoricos(exploratory_data)
exploratory_analysis.head()
```

```python
def identificar_tipos_variables_con_tabla(dataset):
    # Identificar columnas numéricas
    variables_numericas = dataset.select_dtypes(include=['number']).columns.tolist()

    # Identificar columnas categóricas
    variables_categoricas = dataset.select_dtypes(include=['object', 'category']).columns.tolist()

    # Crear un DataFrame para mostrar la información en formato de tabla
    data = {
        "Variables Numéricas": pd.Series(variables_numericas),
        "Variables Categóricas": pd.Series(variables_categoricas)
    }

    tabla_variables = pd.DataFrame(data)

    return tabla_variables

# Aplicar la función en tu dataset 'clean_IBM'
tabla_variables = identificar_tipos_variables_con_tabla(exploratory_analysis)

# Mostrar la tabla
print(tabla_variables)
```

**Prueba de Chi-cuadrado para Identificar Relaciones Significativas**

La primera fase consiste en aplicar la prueba de Chi-cuadrado para identificar las relaciones estadísticamente significativas entre pares de variables categóricas. El test permite determinar si existe una asociación entre las variables y si las diferencias observadas en los datos son lo suficientemente grandes como para no ser atribuidas al azar.

Pasos:

* Aplicar el test de Chi-cuadrado a las combinaciones de variables categóricas.
* Filtrar las relaciones con un p-valor < 0.05, ya que estas se consideran significativas.
* Ordenar los resultados por el p-valor, para priorizar el análisis de las relaciones más relevantes.

Consideración:

* Si el p-valor es bajo (por ejemplo, p < 0.05), se concluye que existe una relación significativa entre las variables, lo que podría ser indicativo de redundancia si ambas variables están aportando información similar.

```python
# 1. Prueba de Chi-cuadrado para Identificar Relaciones Significativas con Conclusión
def chi2_test(df, categorical_columns, alpha=0.05):
    chi2_results = []

    for i in range(len(categorical_columns)):
        for j in range(i+1, len(categorical_columns)):
            var1 = categorical_columns[i]
            var2 = categorical_columns[j]

            # Crear tabla de contingencia
            contingency_table = pd.crosstab(df[var1], df[var2])

            # Realizar el test de Chi-cuadrado
            chi2, p, dof, expected = chi2_contingency(contingency_table)

            # Conclusión sobre si es significativa
            conclusion = 'Significativa' if p < alpha else 'No Significativa'

            # Almacenar el resultado con conclusión
            chi2_results.append({
                'Variable 1': var1,
                'Variable 2': var2,
                'Chi2': chi2,
                'p-value': p,
                'DoF': dof,
                'Conclusión': conclusion
            })

    # Ordenar los resultados por p-valor
    chi2_results_df = pd.DataFrame(chi2_results).sort_values(by='p-value')
    return chi2_results_df

```

```python
# Ejecución del análisis
# Definir las columnas categóricas (ajustar según el dataset)
categorical_columns = ['Attrition', 'BusinessTravel', 'Department', 'EducationField',
                       'Gender', 'JobRole', 'MaritalStatus', 'OverTime']

# Suponiendo que tu dataset ya está cargado en la variable df_nominal
df_nominal = exploratory_analysis  # Cargar el dataset con variables categóricas

# Paso 1: Prueba de Chi-cuadrado
chi2_results_df = chi2_test(df_nominal, categorical_columns)
chi2_results_df
```

**Uso de V de Cramer para Cuantificar la Fuerza de la Relación**

Una vez identificadas las relaciones significativas con la prueba de Chi-cuadrado, se emplea el V de Cramer para medir la fuerza de la relación. Esta métrica ofrece una visión cuantitativa sobre cuán asociadas están las variables, lo que ayuda a determinar si la relación es suficientemente fuerte como para considerar la eliminación de una de las variables por redundancia.

Pasos:

* Calcular el valor de V de Cramer para las relaciones previamente identificadas como significativas.
* Interpretar el valor de V de Cramer en función de su magnitud:
  * V > 0.3: Relación fuerte.
  * 0.1 < V < 0.3: Relación moderada.
  * V < 0.1: Relación débil.

Consideración:

* Las relaciones con un V de Cramer alto (V > 0.3) indican una asociación fuerte, lo que puede ser un indicio claro de redundancia si ambas variables tienen un contexto similar.
* Incluso si las variables están fuertemente asociadas, es importante considerar su contexto antes de eliminarlas. Si aportan valor único o tienen roles diferenciados en el análisis, podrían ser relevantes por separado.

```python
# 2. Uso de V de Cramer para Cuantificar la Fuerza de la Relación

# Función para calcular el V de Cramer
def cramers_v(contingency_table):
    chi2, p, dof, expected = chi2_contingency(contingency_table)
    n = contingency_table.sum().sum()
    return np.sqrt(chi2 / (n * (min(contingency_table.shape) - 1)))

# 2. Uso de V de Cramer para Cuantificar la Fuerza de la Relación y Mapa de Calor
def calculate_cramers_v(df, chi2_results_df):
    cramers_v_results = []
    cramers_v_matrix = pd.DataFrame(np.zeros((len(categorical_columns), len(categorical_columns))),
                                    index=categorical_columns, columns=categorical_columns)

    for _, row in chi2_results_df.iterrows():
        var1 = row['Variable 1']
        var2 = row['Variable 2']

        # Crear tabla de contingencia
        contingency_table = pd.crosstab(df[var1], df[var2])

        # Calcular V de Cramer
        v = cramers_v(contingency_table)

        # Rellenar la matriz de V de Cramer
        cramers_v_matrix.loc[var1, var2] = v
        cramers_v_matrix.loc[var2, var1] = v

        # Clasificación según la magnitud de V de Cramer
        if v > 0.3:
            strength = 'Fuerte'
        elif 0.1 < v <= 0.3:
            strength = 'Moderada'
        else:
            strength = 'Débil'

        cramers_v_results.append({
            'Variable 1': var1,
            'Variable 2': var2,
            'V de Cramer': v,
            'Fuerza de Relación': strength
        })

    # Visualizar el mapa de calor de V de Cramer
    plt.figure(figsize=(10, 8))
    sns.heatmap(cramers_v_matrix, annot=True, cmap='coolwarm', cbar=True)
    plt.title('Matriz de V de Cramer para Variables Nominales')
    plt.show()

    cramers_v_df = pd.DataFrame(cramers_v_results)
    return cramers_v_df
```

**Análisis de Influencia de Variables en Attrition**

Este análisis se centra en identificar las variables con influencia significativa en la probabilidad de Attrition (deserción) en función de sus relaciones con esta variable objetivo. Para ello, abordaremos las variables numéricas, ordinales y nominales, utilizando pruebas estadísticas y visualizaciones adecuadas.

Variables numericas:

Se aplica la Prueba de Mann-Whitney U y Coeficiente de Correlación de Spearman:

1. Prueba de Mann-Whitney U

  Objetivo: Dado que las variables numéricas no cumplen con la normalidad, usaremos la prueba no paramétrica de Mann-Whitney U para comparar las distribuciones de cada variable numérica entre Attrition=Yes y Attrition=No.

  Si el p-valor es significativo (por ejemplo, p ≤ 0.05), podemos concluir que existen diferencias estadísticamente significativas en la distribución de la variable numérica entre los dos grupos de Attrition.
2. Coeficiente de Correlación de Spearman

  Objetivo: Calcularemos la correlación de Spearman entre cada variable numérica y Attrition, codificando Attrition como 0 (No) y 1 (Yes). Esto nos permitirá evaluar la dirección y fuerza de la relación de cada variable numérica con Attrition.
  
  Un valor positivo indica una relación directa (a mayores valores de la variable numérica, aumenta la probabilidad de Attrition=Yes), mientras que un valor negativo indica una relación inversa.

```python
# Lista de variables numéricas
num_vars = ['Age', 'DailyRate', 'DistanceFromHome', 'HourlyRate', 'MonthlyIncome',
            'MonthlyRate', 'NumCompaniesWorked', 'PercentSalaryHike', 'StockOptionLevel',
            'TotalWorkingYears', 'TrainingTimesLastYear', 'YearsAtCompany',
            'YearsInCurrentRole', 'YearsSinceLastPromotion', 'YearsWithCurrManager']

# Crear DataFrame para almacenar los resultados
numeric_analysis_results = []

# Codificar Attrition para la correlación de Spearman
df_balanced['Attrition_numeric'] = df_balanced['Attrition'].apply(lambda x: 1 if x == 'Yes' else 0)

# Realizar las pruebas de Mann-Whitney U y Spearman para cada variable numérica
for var in num_vars:
    # Mann-Whitney U Test
    stat, p_value_mannwhitney = mannwhitneyu(df_balanced[df_balanced['Attrition'] == 'Yes'][var],
                                             df_balanced[df_balanced['Attrition'] == 'No'][var])

    # Spearman Correlation
    spearman_corr, p_value_spearman = spearmanr(df_balanced[var], df_balanced['Attrition_numeric'])

    # Determinar si hay diferencias significativas
    significance_mannwhitney = "Significativa" if p_value_mannwhitney <= 0.05 else "No significativa"
    significance_spearman = "Significativa" if p_value_spearman <= 0.05 else "No significativa"

    # Almacenar los resultados en el DataFrame
    numeric_analysis_results.append({
        'Variable': var,
        'P-valor Mann-Whitney U': p_value_mannwhitney,
        'Diferencia Mann-Whitney U': significance_mannwhitney,
        'Spearman Correlation': spearman_corr,
        'P-valor Spearman': p_value_spearman,
        'Diferencia Spearman': significance_spearman
    })

# Convertir los resultados a un DataFrame
numeric_analysis_df = pd.DataFrame(numeric_analysis_results)
print("Resultados de Prueba Mann-Whitney U y Spearman para Variables Numéricas:")
numeric_analysis_df
```

Variable o caracterisitcas nominales:

Objetivo: Evaluar la asociación entre las variables nominales y Attrition, midiendo también la intensidad de la relación.

Pruebas Utilizadas:

* Prueba Chi-cuadrado: Evalúa la asociación entre Attrition y cada variable nominal.

* Cramer’s V: Mide la intensidad de la relación entre Attrition y cada variable nominal, complementando la prueba Chi-cuadrado.

  - 0.00 a 0.10: Asociación muy débil o inexistente.
  - 0.10 a 0.20: Asociación débil.
  - 0.20 a 0.30: Asociación moderada.
  - '> 0.30: Asociación fuerte.

  Un valor más alto indica una relación más fuerte entre la variable nominal y Attrition


```python
nominal_vars = ['BusinessTravel', 'EducationField', 'Gender',
                'JobRole', 'MaritalStatus', 'OverTime']

chi2_cramersv_results = []

for var in nominal_vars:
    # Crear tabla de contingencia
    contingency_table = pd.crosstab(df_balanced[var], df_balanced['Attrition'])

    # Prueba Chi-cuadrado
    stat, p_value, _, _ = chi2_contingency(contingency_table)

    # Cálculo de Cramer's V
    n = contingency_table.sum().sum()  # Tamaño de la muestra
    cramer_v = np.sqrt(stat / (n * (min(contingency_table.shape) - 1)))

    # Guardar resultados
    significance = "Significativa" if p_value <= 0.05 else "No significativa"
    chi2_cramersv_results.append({'Variable': var, 'P-valor': p_value, 'Diferencia': significance, 'Cramer’s V': cramer_v})

# Crear DataFrame de resultados
chi2_cramersv_df = pd.DataFrame(chi2_cramersv_results)
print("Resultados de Chi-cuadrado y Cramer's V para Variables Nominales:")
chi2_cramersv_df
```

Conversión a Escala Numérica:

```python

# Definir los mapeos para las variables ordinales
ordinal_mappings = {
    'Education': {'Below College': 1, 'College': 2, 'Bachelor': 3, 'Master': 4, 'Doctor': 5},
    'EnvironmentSatisfaction': {'Low': 1, 'Medium': 2, 'High': 3, 'Very High': 4},
    'JobInvolvement': {'Low': 1, 'Medium': 2, 'High': 3, 'Very High': 4},
    'JobLevel': {'Entry Level': 1, 'Junior Level': 2, 'Mid Level': 3, 'Senior Level': 4, 'Executive Level': 5},
    'JobSatisfaction': {'Low': 1, 'Medium': 2, 'High': 3, 'Very High': 4},
    'PerformanceRating': {'Excellent': 1, 'Outstanding': 2},
    'RelationshipSatisfaction': {'Low': 1, 'Medium': 2, 'High': 3, 'Very High': 4},
    'WorkLifeBalance': {'Bad': 1, 'Good': 2, 'Better': 3, 'Best': 4}
}

# Crear DataFrame para almacenar resultados
ordinal_results = []

# Aplicar el análisis para cada variable ordinal
for var, mapping in ordinal_mappings.items():
    # Convertir la variable ordinal a escala numérica
    df_balanced[f'{var}_numeric'] = df_balanced[var].map(mapping)

    # Crear tabla de contingencia para la prueba Chi-cuadrado
    contingency_table = pd.crosstab(df_balanced[f'{var}_numeric'], df_balanced['Attrition'])
    stat, p_value_chi2, _, _ = chi2_contingency(contingency_table)

    # Calcular correlación de Spearman
    spearman_corr, p_value_spearman = spearmanr(df_balanced[f'{var}_numeric'], df_balanced['Attrition_numeric'])

    # Guardar resultados
    significance_chi2 = "Significativa" if p_value_chi2 <= 0.05 else "No significativa"
    significance_spearman = "Significativa" if p_value_spearman <= 0.05 else "No significativa"

    ordinal_results.append({
        'Variable': var,
        'P-valor Chi2': p_value_chi2,
        'Diferencia Chi2': significance_chi2,
        'Spearman Correlation': spearman_corr,
        'P-valor Spearman': p_value_spearman,
        'Diferencia Spearman': significance_spearman
    })

# Crear DataFrame de resultados
ordinal_df = pd.DataFrame(ordinal_results)
print("Resultados de Chi-cuadrado de Tendencia y Correlación de Spearman para Variables Ordinales:")
ordinal_df

```

```python

def plot_all_numeric_variables(data, numeric_vars, target_var='Attrition'):
    """
    Genera una matriz de gráficos para cada variable numérica en la lista.
    La matriz tiene tres columnas: Boxplot, Violin plot y Histograma.
    Los gráficos se colorean según el valor de la variable objetivo.
    En el boxplot, se añaden la media y mediana, y en el violin plot, se añade la moda para cada clase.

    Parámetros:
    - data: DataFrame de pandas con los datos
    - numeric_vars: lista de nombres de variables numéricas a analizar (list de str)
    - target_var: nombre de la variable objetivo, por defecto 'Attrition'

    """
    num_vars = len(numeric_vars)
    fig, axes = plt.subplots(num_vars, 3, figsize=(18, 6 * num_vars))

    for idx, numeric_var in enumerate(numeric_vars):
        # Boxplot con media y mediana
        ax_box = axes[idx, 0]
        sns.boxplot(data=data, x=target_var, y=numeric_var, palette="Set2", ax=ax_box)
        ax_box.set_title(f"Boxplot de {numeric_var} según {target_var}")

        # Calcular y agregar media y mediana para cada clase de Attrition
        for class_value in data[target_var].unique():
            class_data = data[data[target_var] == class_value][numeric_var]
            mean_val = class_data.mean()
            median_val = class_data.median()
            ax_box.text(x=list(data[target_var].unique()).index(class_value), y=mean_val,
                        s=f'Media: {mean_val:.2f}', color='blue', ha='center', va='bottom')
            ax_box.text(x=list(data[target_var].unique()).index(class_value), y=median_val,
                        s=f'Mediana: {median_val:.2f}', color='green', ha='center', va='top')

        # Violin plot con moda
        ax_violin = axes[idx, 1]
        sns.violinplot(data=data, x=target_var, y=numeric_var, palette="Set2", ax=ax_violin)
        ax_violin.set_title(f"Violin plot de {numeric_var} según {target_var}")

        # Calcular y agregar moda para cada clase de Attrition
        for class_value in data[target_var].unique():
            class_data = data[data[target_var] == class_value][numeric_var]
            mode_val = class_data.mode()[0]
            ax_violin.text(x=list(data[target_var].unique()).index(class_value), y=mode_val,
                           s=f'Moda: {mode_val:.2f}', color='red', ha='center', va='bottom')

        # Histograma con línea de densidad solamente
        ax_hist = axes[idx, 2]
        sns.histplot(data=data, x=numeric_var, hue=target_var, palette="Set2", kde=True, ax=ax_hist)

        # Aumentar el grosor de la línea de densidad
        for line in ax_hist.lines:
            line.set_linewidth(2.5)

        ax_hist.set_title(f"Histograma de {numeric_var} con distribución de densidad")

    plt.tight_layout()
    plt.show()

# Uso de la función con la variable adicional 'StockOptionLevel'
numeric_variables = [
                      "Age", "MonthlyIncome", "StockOptionLevel", "TotalWorkingYears",
        "YearsAtCompany", "YearsInCurrentRole", "YearsWithCurrManager",
        "DailyRate", "HourlyRate", "NumCompaniesWorked", "YearsSinceLastPromotion"]
plot_all_numeric_variables(data=df_balanced, numeric_vars=numeric_variables, target_var='Attrition')
```

```python


def plot_all_nominal_variables(data, nominal_vars, target_var='Attrition'):
    """
    Genera gráficos para cada variable nominal en la lista.
    La matriz tiene tres columnas: Gráfico de Barras, Diagrama de Pareto y Heatmap de Contingencia.
    Los gráficos se colorean según el valor de la variable objetivo.

    Parámetros:
    - data: DataFrame de pandas con los datos
    - nominal_vars: lista de nombres de variables nominales a analizar (list de str)
    - target_var: nombre de la variable objetivo, por defecto 'Attrition'

    """
    num_vars = len(nominal_vars)
    fig, axes = plt.subplots(num_vars, 3, figsize=(18, 6 * num_vars))

    for idx, nominal_var in enumerate(nominal_vars):
        # Calcular el orden de las categorías basado en el Diagrama de Pareto
        pareto_data = data[nominal_var].value_counts().sort_values(ascending=False)
        ordered_categories = pareto_data.index

        # Gráfico de Barras Apilado con orden de categorías del Pareto
        ax_bar = axes[idx, 0]
        sns.countplot(data=data, x=nominal_var, hue=target_var, order=ordered_categories, palette="Set2", ax=ax_bar)
        ax_bar.set_title(f"Gráfico de Barras de {nominal_var} según {target_var}")
        ax_bar.set_xticklabels(ax_bar.get_xticklabels(), rotation=45, ha='right')

        # Añadir línea que una las barras por clase
        for class_value in data[target_var].unique():
            class_counts = data[data[target_var] == class_value][nominal_var].value_counts().reindex(ordered_categories).fillna(0)
            ax_bar.plot(range(len(ordered_categories)), class_counts, marker='o', linestyle='-', label=f'{class_value}')
        ax_bar.legend()

        # Diagrama de Pareto para todas las variables
        ax_pareto = axes[idx, 1]
        pareto_cumsum = pareto_data.cumsum() / pareto_data.sum() * 100  # Porcentaje acumulado

        # Graficar barras de Pareto
        sns.barplot(x=ordered_categories, y=pareto_data.values, ax=ax_pareto, palette="Set2")
        ax_pareto.set_ylabel("Frecuencia")
        ax_pareto.set_title(f"Diagrama de Pareto de {nominal_var}")
        ax_pareto.set_xticklabels(ax_pareto.get_xticklabels(), rotation=45, ha='right')

        # Añadir línea acumulada de porcentaje
        ax_pareto_2 = ax_pareto.twinx()
        ax_pareto_2.plot(ordered_categories, pareto_cumsum, color="blue", marker="o", linestyle="-", linewidth=2)
        ax_pareto_2.set_ylabel("Frecuencia Acumulada (%)")
        ax_pareto_2.grid(False)

        # Heatmap de Contingencia
        ax_heatmap = axes[idx, 2]
        contingency_table = pd.crosstab(data[nominal_var], data[target_var])
        sns.heatmap(contingency_table, annot=True, fmt='d', cmap="YlGnBu", ax=ax_heatmap)
        ax_heatmap.set_title(f"Heatmap de Contingencia de {nominal_var} y {target_var}")

    plt.tight_layout()
    plt.show()

# Ejemplo de uso
nominal_variables = ["BusinessTravel", "JobRole", "MaritalStatus", "OverTime"]
plot_all_nominal_variables(data=df_balanced, nominal_vars=nominal_variables, target_var='Attrition')
```

```python


def plot_all_ordinal_variables(data, ordinal_vars, target_var='Attrition', mappings=None):
    """
    Genera gráficos para cada variable ordinal en la lista.
    La matriz tiene tres columnas: Gráfico de Barras Apilado, Boxplot y Violin Plot.
    Los gráficos se colorean según el valor de la variable objetivo y respetan el orden de los niveles ordinales.

    Parámetros:
    - data: DataFrame de pandas con los datos
    - ordinal_vars: lista de nombres de variables ordinales a analizar (list de str)
    - target_var: nombre de la variable objetivo, por defecto 'Attrition'
    - mappings: diccionario con los mapeos de valores numéricos a categorías ordinales

    """
    num_vars = len(ordinal_vars)
    fig, axes = plt.subplots(num_vars, 3, figsize=(18, 6 * num_vars))

    for idx, ordinal_var in enumerate(ordinal_vars):
        # Verificar si la variable tiene un mapeo en 'mappings' y aplicar el mapeo si existe
        if mappings and ordinal_var in mappings:
            inverse_mapping = {v: k for k, v in mappings[ordinal_var].items()}
            data[f'{ordinal_var}_numeric'] = data[ordinal_var].map(inverse_mapping)
            category_order = list(mappings[ordinal_var].values())
        else:
            # Si no hay mapeo, asumir valores numéricos ya en la variable
            data[f'{ordinal_var}_numeric'] = data[ordinal_var]
            category_order = sorted(data[ordinal_var].unique())

        # Gráfico de Barras Apilado con nombres de categorías originales
        ax_bar = axes[idx, 0]
        sns.countplot(data=data, x=ordinal_var, hue=target_var, order=category_order, palette="Set2", ax=ax_bar)
        ax_bar.set_title(f"Gráfico de Barras de {ordinal_var} según {target_var}")
        ax_bar.tick_params(axis='x', rotation=45)

        # Añadir línea que una las barras por clase
        for class_value in data[target_var].unique():
            class_counts = data[data[target_var] == class_value][ordinal_var].value_counts().reindex(category_order).fillna(0)
            ax_bar.plot(range(len(category_order)), class_counts, marker='o', linestyle='-', label=f'{class_value}')
        ax_bar.legend()

        # Boxplot con media y mediana para cada clase usando valores numéricos temporales
        ax_box = axes[idx, 1]
        sns.boxplot(data=data, x=target_var, y=f'{ordinal_var}_numeric', palette="Set2", ax=ax_box)
        ax_box.set_title(f"Boxplot de {ordinal_var} según {target_var}")
        ax_box.set_yticks(range(1, len(category_order) + 1))  # Configura el eje Y con las categorías
        ax_box.set_yticklabels(category_order)  # Etiquetas de categorías originales

        # Calcular y agregar media y mediana para cada clase
        for class_value in data[target_var].unique():
            class_data = data[data[target_var] == class_value][f'{ordinal_var}_numeric']
            mean_val = class_data.mean()
            median_val = class_data.median()
            ax_box.text(x=list(data[target_var].unique()).index(class_value), y=mean_val,
                        s=f'Media: {mean_val:.2f}', color='blue', ha='center', va='bottom')
            ax_box.text(x=list(data[target_var].unique()).index(class_value), y=median_val,
                        s=f'Mediana: {median_val:.2f}', color='green', ha='center', va='top')

        # Violin plot con moda usando valores numéricos temporales
        ax_violin = axes[idx, 2]
        sns.violinplot(data=data, x=target_var, y=f'{ordinal_var}_numeric', palette="Set2", ax=ax_violin)
        ax_violin.set_title(f"Violin plot de {ordinal_var} según {target_var}")
        ax_violin.set_yticks(range(1, len(category_order) + 1))  # Configura el eje Y con las categorías
        ax_violin.set_yticklabels(category_order)  # Etiquetas de categorías originales

        # Calcular y agregar moda para cada clase
        for class_value in data[target_var].unique():
            class_data = data[data[target_var] == class_value][f'{ordinal_var}_numeric']
            mode_val = class_data.mode()[0]
            ax_violin.text(x=list(data[target_var].unique()).index(class_value), y=mode_val,
                           s=f'Moda: {mode_val:.2f}', color='red', ha='center', va='bottom')

        # Eliminar la columna temporal después de los gráficos
        data.drop(columns=[f'{ordinal_var}_numeric'], inplace=True)

    plt.tight_layout()
    plt.show()

# Diccionario de mappings proporcionado
mappings = {
    'Education': {1: 'Below College', 2: 'College', 3: 'Bachelor', 4: 'Master', 5: 'Doctor'},
    'EnvironmentSatisfaction': {1: 'Low', 2: 'Medium', 3: 'High', 4: 'Very High'},
    'JobInvolvement': {1: 'Low', 2: 'Medium', 3: 'High', 4: 'Very High'},
    'JobSatisfaction': {1: 'Low', 2: 'Medium', 3: 'High', 4: 'Very High'},
    'PerformanceRating': {1: 'Low', 2: 'Good', 3: 'Excellent', 4: 'Outstanding'},
    'RelationshipSatisfaction': {1: 'Low', 2: 'Medium', 3: 'High', 4: 'Very High'},
    'WorkLifeBalance': {1: 'Bad', 2: 'Good', 3: 'Better', 4: 'Best'},
    'JobLevel': {1: "Entry Level", 2: "Junior Level", 3: "Mid Level", 4: "Senior Level", 5: "Executive Level"}
}

# Ejemplo de uso
ordinal_variables = ["JobInvolvement", "JobLevel", "JobSatisfaction"]
plot_all_ordinal_variables(data=df_balanced, ordinal_vars=ordinal_variables, target_var='Attrition', mappings=mappings)
```

```python
def evaluar_entrenamiento(model, X_train, y_train):
    y_train_pred = model.predict(X_train)

    print("Resultados de entrenamiento: \n")
    clf_report = pd.DataFrame(classification_report(y_train, y_train_pred, output_dict=True))
    print(f"Matriz de confusión:\n{confusion_matrix(y_train, y_train_pred)}")
    print(f"Resultados accuracy:\n{accuracy_score(y_train, y_train_pred):.4f}")
    print(f"Reporte de clasificación:\n{clf_report}")
```

```python
def plot_precision_recall_vs_threshold(precisions, recalls, thresholds):
    plt.plot(thresholds, precisions[:-1], "b--", label="Precision")
    plt.plot(thresholds, recalls[:-1], "g--", label="Recall")
    plt.xlabel("Umbral")
    plt.legend(loc="upper left")
    plt.title("Compensación entre Precisión y Recall")


def plot_roc_curve(fpr, tpr, label=None):
    plt.plot(fpr, tpr, linewidth=2, label=label)
    plt.plot([0, 1], [0, 1], "k--")
    plt.axis([0, 1, 0, 1])
    plt.xlabel('Tasa de Falsos Positivos')
    plt.ylabel('Tasa de Verdaderos Positivos')
    plt.title('Curva ROC')
```

**1. Curva de Validación: Evaluar el Impacto del Hiperparámetro C**

El parámetro C en regresión logística controla la regularización. Se probarán una variedad de valores para C en escala logarítmica para ver cómo afecta el rendimiento en entrenamiento y validación.

```python
# Rango de valores para el hiperparámetro C
param_range = np.logspace(-4, 4, 10)  # Valores de C en escala logarítmica

# Generar la curva de validación
train_scores, val_scores = validation_curve(
    LogisticRegression(class_weight='balanced', solver='liblinear', penalty='l1', n_jobs=-1, random_state=42),
    X_train, y_train, param_name="C", param_range=param_range,
    cv=5, scoring="accuracy", n_jobs=-1
)

# Calcular medias y desviaciones estándar
train_scores_mean = np.mean(train_scores, axis=1)
train_scores_std = np.std(train_scores, axis=1)
val_scores_mean = np.mean(val_scores, axis=1)
val_scores_std = np.std(val_scores, axis=1)

# Graficar la curva de validación
plt.figure(figsize=(10, 6))
plt.plot(param_range, train_scores_mean, label="Puntaje de entrenamiento", color="blue")
plt.plot(param_range, val_scores_mean, label="Puntaje de validación", color="orange")
plt.fill_between(param_range, train_scores_mean - train_scores_std,
                 train_scores_mean + train_scores_std, color="blue", alpha=0.2)
plt.fill_between(param_range, val_scores_mean - val_scores_std,
                 val_scores_mean + val_scores_std, color="orange", alpha=0.2)
plt.xscale("log")
plt.xlabel("C (Regularización)")
plt.ylabel("Puntaje")
plt.title("Curva de Validación para el Hiperparámetro C")
plt.legend(loc="best")
plt.show()
```

𝐶 en la regresión logística nos proporciona una visión del efecto de la regularización en el rendimiento del modelo en los conjuntos de entrenamiento y validación.

* A medida que 𝐶 aumenta, el puntaje de entrenamiento y validación también aumentan, indicando una mejora en la capacidad del modelo para ajustarse a los datos.

## 6.1 Dividir el Conjunto de Datos en Entrenamiento, Validación y Prueba (ya realizado)

* Conjunto de Entrenamiento: Usado para ajustar el modelo.
* Conjunto de Validación: Usado para seleccionar los mejores hiperparámetros y evitar el sobreajuste.

La idea es que el conjunto de prueba actúe como una representación de datos futuros o no observados, lo cual permite una evaluación más realista de la capacidad del modelo para generalizar. En general, los datos pueden dividirse de la siguiente manera:

* 70% para entrenamiento
* 15% para validación
* 15% para prueba
